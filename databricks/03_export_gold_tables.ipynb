```python
from pyspark.sql import functions as F

GOLD_TABLES = [
    "gold_passenger_journeys",
    "gold_daily_journey_summary",
    "gold_hourly_demand_summary",
    "gold_origin_destination_summary",
    "gold_passenger_category_summary",
    "gold_payment_method_summary",
    "gold_route_summary",
    "gold_time_period_summary",
    "gold_transport_mode_summary"
]

EXPORT_BASE = "/Volumes/workspace/default/dublin_transport_gtfs/gold_exports"

print("Gold tables to export:", len(GOLD_TABLES))
print("Export location:", EXPORT_BASE)
```

```python

import os
import pandas as pd

os.makedirs(EXPORT_BASE, exist_ok=True)

export_results = []

for table_name in GOLD_TABLES:
    source_table = f"workspace.default.{table_name}"
    output_file = f"{EXPORT_BASE}/{table_name}.csv"

    df = spark.table(source_table)
    source_count = df.count()

    # Suitable for the small, validated Gold datasets
    df.toPandas().to_csv(output_file, index=False)

    # Verify exported CSV row count
    exported_count = len(pd.read_csv(output_file))

    assert source_count == exported_count, (
        f"Row count mismatch: {table_name}"
    )

    export_results.append(
        (table_name, source_count, exported_count, "PASS")
    )

    print(f"Exported: {table_name} ({source_count} rows)")

display(
    spark.createDataFrame(
        export_results,
        ["table_name", "source_rows", "exported_rows", "status"]
    )
)
```
